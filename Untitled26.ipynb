```python

!pip install -q gradio tensorflow pillow numpy

import os
import numpy as np
import gradio as gr
import tensorflow as tf
from PIL import Image, ImageOps

# =====================================================
# SMART WASTE MANAGEMENT WITH COMPUTER VISION
# =====================================================

APP_TITLE = "Smart Waste Management"

# Set the path to your trained 8-class model if available.
# Example: MODEL_PATH = "/content/waste_model.keras"
MODEL_PATH = ""

CLASS_NAMES = [
    "cardboard", "glass", "metal", "paper",
    "plastic", "trash", "organic", "hazardous"
]

# -----------------------------------------------------
# WASTE CATEGORY DATABASE
# -----------------------------------------------------

WASTE_GUIDE = {
    "cardboard": {
        "icon": "📦",
        "category": "Recyclable Waste",
        "bin": "Dry waste / recycling",
        "advice": "Flatten cardboard boxes. Keep them clean and dry.",
        "description": "Boxes, packaging and cardboard sheets."
    },
    "glass": {
        "icon": "🍾",
        "category": "Recyclable Waste",
        "bin": "Glass collection",
        "advice": "Separate glass bottles and jars. Handle broken glass carefully.",
        "description": "Glass bottles and jars accepted by local recycling services."
    },
    "metal": {
        "icon": "🥫",
        "category": "Recyclable Waste",
        "bin": "Metal recycling",
        "advice": "Empty cans and containers before recycling.",
        "description": "Metal cans and recyclable metal containers."
    },
    "paper": {
        "icon": "📄",
        "category": "Recyclable Waste",
        "bin": "Paper recycling",
        "advice": "Keep paper clean and dry. Greasy paper may not be recyclable.",
        "description": "Newspapers, office paper and suitable paper packaging."
    },
    "plastic": {
        "icon": "♻️",
        "category": "Recyclable Where Accepted",
        "bin": "Plastic collection",
        "advice": "Check the plastic type and local recycling rules.",
        "description": "Plastic bottles, containers and packaging."
    },
    "trash": {
        "icon": "🗑️",
        "category": "Residual Waste",
        "bin": "Residual waste collection",
        "advice": "Check whether the item can be reused or recycled first.",
        "description": "Waste that cannot currently be reused or recycled."
    },
    "organic": {
        "icon": "🌱",
        "category": "Organic / Biodegradable Waste",
        "bin": "Wet waste / composting",
        "advice": "Separate food scraps and suitable garden waste for composting.",
        "description": "Fruit peels, vegetable scraps and food leftovers."
    },
    "hazardous": {
        "icon": "⚠️",
        "category": "Hazardous / Special Waste",
        "bin": "Authorized special collection",
        "advice": "Keep batteries, chemicals, e-waste and medical waste out of ordinary bins.",
        "description": "Potentially harmful materials requiring special handling."
    }
}

# -----------------------------------------------------
# LOAD MODEL
# -----------------------------------------------------

model = None
model_error = None

if MODEL_PATH.strip():
    if os.path.exists(MODEL_PATH):
        try:
            model = tf.keras.models.load_model(MODEL_PATH)
            print("AI model loaded successfully.")
        except Exception as error:
            model_error = str(error)
            print("Model loading failed:", error)
    else:
        model_error = "The model file was not found."

if model is None:
    print("No trained model configured. Manual category selection is available.")

# -----------------------------------------------------
# CATEGORY CARDS
# -----------------------------------------------------

def create_category_cards():
    cards = ""

    for key, item in WASTE_GUIDE.items():
        cards += f"""
        <div class="category-card">
            <div class="category-icon">{item['icon']}</div>
            <h3>{key.title()}</h3>
            <p>{item['description']}</p>
            <div class="bin-label">Recommended disposal</div>
            <p>{item['bin']}</p>
            <div class="advice">{item['advice']}</div>
        </div>
        """

    return f'<div class="category-grid">{cards}</div>'

# -----------------------------------------------------
# IMAGE PREPROCESSING
# -----------------------------------------------------

def preprocess_image(image, height, width):
    image = ImageOps.exif_transpose(image).convert("RGB")
    image = image.resize((width, height))
    array = np.asarray(image, dtype=np.float32) / 255.0
    return np.expand_dims(array, axis=0)

# -----------------------------------------------------
# CLASSIFICATION
# -----------------------------------------------------

def classify_waste(image):
    if image is None:
        return (
            "No image uploaded",
            "Please upload an image or select a category manually.",
            "No recommendation available.",
            None
        )

    image = ImageOps.exif_transpose(image).convert("RGB")

    if model is None:
        return (
            "AI prediction unavailable: no trained model configured.",
            "Choose a category from the manual category dropdown below.",
            "Manual selection does not automatically identify the image.",
            image
        )

    try:
        input_shape = (
            model.input_shape[0]
            if isinstance(model.input_shape, list)
            else model.input_shape
        )

        height = int(input_shape[1] or 224)
        width = int(input_shape[2] or 224)

        predictions = np.asarray(
            model.predict(preprocess_image(image, height, width), verbose=0)
        )

        if predictions.ndim != 2 or predictions.shape[0] != 1:
            raise ValueError("Expected one image-classification output.")

        scores = predictions[0].astype(float)

        if len(scores) != len(CLASS_NAMES):
            raise ValueError(
                f"Model outputs {len(scores)} classes; expected 8."
            )

        if (
            np.any(scores < 0)
            or np.any(scores > 1)
            or not np.isclose(np.sum(scores), 1.0, atol=0.05)
        ):
            scores = np.exp(scores - np.max(scores))
            scores /= np.sum(scores)

        index = int(np.argmax(scores))
        label = CLASS_NAMES[index]
        confidence = float(scores[index]) * 100
        info = WASTE_GUIDE[label]

        prediction = (
            f"{info['icon']} Predicted Waste: {label.title()}\n"
            f"Confidence: {confidence:.2f}%"
        )

        category = (
            f"{info['category']}\n"
            f"Recommended stream: {info['bin']}"
        )

        advice = (
            f"{info['description']}\n\n"
            f"Recommendation: {info['advice']}"
        )

        if confidence < 60:
            prediction += "\n\nLow confidence. Please review manually."

        return prediction, category, advice, image

    except Exception as error:
        return (
            "Classification failed",
            "Prediction unavailable",
            str(error),
            image
        )

# -----------------------------------------------------
# MANUAL CATEGORY SELECTION
# -----------------------------------------------------

def show_category(category_name):
    if not category_name:
        return "No category selected.", "Select a waste category to see its recommendation."

    info = WASTE_GUIDE[category_name]

    category_text = (
        f"{info['icon']} {category_name.title()}\n"
        f"Category: {info['category']}\n"
        f"Recommended stream: {info['bin']}"
    )

    advice_text = (
        f"{info['description']}\n\n"
        f"Recommendation: {info['advice']}\n\n"
        "Note: This is based on your manual selection, not AI image recognition."
    )

    return category_text, advice_text

# -----------------------------------------------------
# WHITE BACKGROUND WEBSITE DESIGN
# -----------------------------------------------------

CUSTOM_CSS = """
@import url('https://fonts.googleapis.com/css2?family=DM+Sans:wght@400;500;600;700&family=Space+Grotesk:wght@500;600;700&display=swap');

.gradio-container {
    max-width: 1250px !important;
    margin: auto !important;
    font-family: 'DM Sans', sans-serif !important;
    background: #ffffff !important;
    color: #1f2937 !important;
}

body, .gradio-container, .main, .contain {
    background: #ffffff !important;
    color: #1f2937 !important;
}

h1, h2, h3, p, label, span {
    color: #1f2937;
}

#hero {
    padding: 35px 25px;
    border-radius: 24px;
    background: linear-gradient(120deg, #f0fdf4, #dcfce7, #ffffff);
    border: 1px solid #bbf7d0;
    margin-bottom: 20px;
}

#hero h1 {
    color: #166534 !important;
    font-family: 'Space Grotesk', sans-serif;
    font-size: clamp(30px, 5vw, 46px);
}

#hero p {
    color: #374151 !important;
    font-size: 16px;
    line-height: 1.7;
}

.pill {
    display: inline-block;
    background: #dcfce7;
    border: 1px solid #86efac;
    color: #166534 !important;
    padding: 7px 13px;
    border-radius: 30px;
    font-size: 12px;
}

.category-grid {
    display: grid;
    grid-template-columns: repeat(2, minmax(0, 1fr));
    gap: 15px;
    margin: 15px 0;
}

.category-card {
    padding: 20px;
    background: #ffffff;
    border: 1px solid #d1d5db;
    border-radius: 18px;
}

.category-icon {
    font-size: 30px;
}

.category-card h3 {
    color: #166534 !important;
    margin: 12px 0;
}

.category-card p {
    color: #374151 !important;
    line-height: 1.6;
}

.bin-label {
    font-size: 12px;
    color: #15803d !important;
    text-transform: uppercase;
    letter-spacing: 1px;
}

.advice, .notice {
    background: #f0fdf4;
    border-left: 4px solid #66d98e;
    border-radius: 10px;
    padding: 14px;
    color: #1f2937 !important;
    line-height: 1.6;
    margin: 12px 0;
}

.gradio-container button.primary {
    background: #69dc91 !important;
    color: #173321 !important;
    font-weight: 700 !important;
    border: none !important;
}

.gradio-container button {
    border-radius: 12px !important;
}

.gradio-container input,
.gradio-container textarea,
.gradio-container select {
    background: #ffffff !important;
    color: #1f2937 !important;
}

footer {
    display: none !important;
}

@media(max-width: 650px) {
    .category-grid {
        grid-template-columns: 1fr;
    }
    #hero {
        padding: 22px 16px;
    }
}
"""

HERO_HTML = """
<div id="hero">
    <span class="pill">AI • COMPUTER VISION • SUSTAINABILITY</span>
    <h1>♻️ Smart Waste Management</h1>
    <p>
        Identify waste categories, explore recycling guidance,
        and discover how computer vision can support smarter
        waste segregation.
    </p>
</div>
"""

# -----------------------------------------------------
# BUILD WEBSITE
# -----------------------------------------------------

with gr.Blocks(
    theme=gr.themes.Base(
        primary_hue="green",
        secondary_hue="emerald",
        neutral_hue="slate"
    ),
    css=CUSTOM_CSS,
    title=APP_TITLE
) as demo:

    gr.HTML(HERO_HTML)

    with gr.Tabs():

        # HOME
        with gr.Tab("🏠 Home"):
            gr.Markdown("## Welcome to Smart Waste Management")

            gr.Markdown("""
            This project demonstrates an AI-based waste management workflow:

            1. Upload an image of waste.
            2. Resize and preprocess the image.
            3. Classify it using a trained CNN model.
            4. Display the predicted waste category.
            5. Recommend a suitable disposal or recycling stream.
            """)

            gr.HTML(
                '<div class="notice"><b>Eight waste categories</b><br>'
                'AI predictions require a compatible trained model. '
                'Manual category selection is also available.</div>'
            )

            gr.HTML(create_category_cards())

        # AI CLASSIFICATION
        with gr.Tab("🤖 AI Classification"):
            gr.Markdown("## Upload and Analyze Waste")

            gr.Markdown(
                "Upload an image for AI classification when a trained model "
                "is configured. Otherwise, use manual category selection."
            )

            with gr.Row():
                with gr.Column():
                    image_input = gr.Image(
                        type="pil",
                        label="Upload Waste Image",
                        sources=["upload", "webcam"]
                    )

                    analyze_button = gr.Button(
                        "✨ Analyze Image",
                        variant="primary"
                    )

                with gr.Column():
                    prediction_output = gr.Textbox(
                        label="AI Prediction",
                        lines=3,
                        interactive=False
                    )

                    category_output = gr.Textbox(
                        label="Waste Category",
                        lines=3,
                        interactive=False
                    )

                    advice_output = gr.Textbox(
                        label="Segregation Recommendation",
                        lines=5,
                        interactive=False
                    )

            preview_output = gr.Image(
                label="Uploaded Image Preview",
                interactive=False
            )

            analyze_button.click(
                fn=classify_waste,
                inputs=[image_input],
                outputs=[
                    prediction_output,
                    category_output,
                    advice_output,
                    preview_output
                ]
            )

            gr.Markdown("### Or Select the Waste Category Manually")

            manual_category = gr.Dropdown(
                choices=CLASS_NAMES,
                label="Select Waste Category",
                value=None,
                interactive=True
            )

            manual_category_output = gr.Textbox(
                label="Selected Category",
                lines=3,
                interactive=False
            )

            manual_advice_output = gr.Textbox(
                label="Manual Disposal Recommendation",
                lines=4,
                interactive=False
            )

            manual_category.change(
                fn=show_category,
                inputs=[manual_category],
                outputs=[
                    manual_category_output,
                    manual_advice_output
                ]
            )

        # WASTE CATEGORIES
        with gr.Tab("♻️ Waste Categories"):
            gr.Markdown("## Waste Category and Disposal Guide")
            gr.Markdown(
                "Explore the eight waste categories and their recommendations."
            )
            gr.HTML(create_category_cards())

        # AI TECHNIQUES
        with gr.Tab("🧠 AI Techniques"):
            gr.Markdown("""
            ## Computer Vision Pipeline

            **1. Image acquisition:** Upload a photograph.

            **2. Image preprocessing:** Convert to RGB, resize and normalize pixels.

            **3. CNN classification:** A trained Convolutional Neural Network predicts the waste class.

            **4. Transfer learning:** MobileNetV2 can be fine-tuned on a waste image dataset.

            **5. Confidence checking:** Low-confidence predictions should be reviewed manually.

            **6. Model evaluation:** Use a test set, confusion matrix, precision, recall and F1-score.

            **7. Future scope:** Add YOLO object detection, IoT smart bins, collection monitoring and waste analytics.
            """)

        # ABOUT
        with gr.Tab("ℹ️ About"):
            gr.Markdown("""
            ## About the Project

            **Project:** Smart Waste Management with Computer Vision

            **Technologies:** Python, TensorFlow, Gradio, NumPy and Pillow

            **AI method:** CNN image classification

            **Objective:** Assist waste classification and provide disposal guidance.

            **Limitations:** Automatic predictions depend on model training, class order and preprocessing. Manual category selection is not AI recognition. Follow local waste regulations, especially for hazardous materials.
            """)

    gr.Markdown(
        "---\n"
        "♻️ Smart Waste Management | AI-Based Computer Vision"
    )

# -----------------------------------------------------
# LAUNCH WEBSITE
# -----------------------------------------------------

demo.launch(share=True, debug=True)
```
